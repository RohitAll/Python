```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('matplotlib_practice_data.csv')
```

```python
rating_counts=df['Category'].value_counts()
plt.pie(rating_counts, labels=rating_counts.index, autopct='%1.1f%%')
plt.title('Category-wise Distribution')
plt.ylabel('')
plt.show()
```

```python
rating_counts=df['Rating'].value_counts()
plt.pie(rating_counts, labels=rating_counts.index, autopct='%1.1f%%')
plt.title('Category-wise Distribution')
plt.show()
```

```python
rating_counts=df['Gender'].value_counts()
plt.pie(rating_counts, labels=rating_counts.index, autopct='%1.1f%%')
plt.title('Category-wise Distribution')
plt.show()
```
